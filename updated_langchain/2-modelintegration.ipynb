#### Streaming And Batch

1. Streaming

Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. 

Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time:

```python
model.invoke("Write me a 200 words paragraph on agentic ai?")

# in this case model first generate the content and then displays it in the output cell. while generating the content user will wait for the model to complete the generation. In case of long content generation, it may take time to generate the content and user will have to wait for the model to complete the generation.
```

```python
model.stream("Write me a 200 words paragraph on agentic ai?")

# output -> <generator object BaseChatModel.stream at 0x0000012D26CC6640>

# this is can generator object, but we want to stream it so use loop.
```

```python
for chunk in model.stream("Write me a 200 words paragraph on Artificial Intelligence"):
    print(chunk.text, end="", flush=True)
```

```python
for chunk in model.stream("Why do parrots have colorful feathers?"):
    print(chunk.text, end="", flush=True)
```

2. Batch

Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
response = model.batch([
    "why do parrots have colorful feathers?",
    "how do airplanes fly?",
    "what is agentic ai?",
])

for res in response:
    print(res)
```

```python
model.batch(
    ["Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"],
    config={
        'max_concurrency': 5, # limit to 5 parrel requests at a time
    }
)
```
