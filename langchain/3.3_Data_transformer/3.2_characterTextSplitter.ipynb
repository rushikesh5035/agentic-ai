#### How to split by character --> character text splitter

- This is the simplest method.
- This splits based on a given character sequence, wich defaults to "\n\n". 
- Chuck length is meansured by number of characters.

1. **How the text is split:** by single character separator.
2. **How the chunk size is measured:** by number of characters

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("sample.txt", encoding="utf8")
docs = loader.load()

docs
```

```python
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator="\n\n",  # separator is the string used to split the text into chunks
    chunk_size=100,
    chunk_overlap=20
)

text_splitter.split_documents(docs)
```

```python
text = ""
with open("sample.txt", "r", encoding="utf8") as f:
    text = f.read()

text_splitter = CharacterTextSplitter(chunk_size=100, chunk_overlap=20)
final_texts = text_splitter.create_documents([text])

print(final_texts[0])

if len(final_texts) > 1:
    print(final_texts[1])
```
