```python
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
# Set the Google API key from the environment variable

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")
```

```python
text = "This is a practical example on GOOGLE embedding"

query_result = embeddings.embed_query(text) 

query_result
```

```python
len(query_result) # 3072 dimensions for the embedding vector
```

```python
## reduced-dimensions

from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings_1536 = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview", output_dimensionality=1536)

text = "This is a practical example on GOOGLE embedding"

query_result = embeddings_1536.embed_query(text)

len(query_result) # 1536 dimensions for the embedding vector
```

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("sample.txt", encoding="utf8")
docs = loader.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)

final_documents = text_splitter.split_documents(docs)
final_documents
```

```python
from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(final_documents, embeddings_1536)

db
```

```python
# Retrieve the most similar documents based on a query using similarity search

query = "The experts will enjoy the same treatment as their counterparts from State-owned firms, the report said."

retrieved_result = db.similarity_search(query)

print(retrieved_result)
```
