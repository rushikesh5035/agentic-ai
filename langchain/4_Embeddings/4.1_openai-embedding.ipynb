#### Embedding Techniques

Converting text into vectors

```python
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
# Set the OpenAI API key from the environment variable

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(
    model="text-embedding-3-large"
)
```

```python
text = "This is a practical example on OPENAI embedding"

query_result = embeddings.embed_query(text) # this will return a list of floats representing the embedding vector for the input text

query_result
```

```python
len(query_result) # 3072 -> the length of the embedding vector for the "text-embedding-3-large" model is 3072 dimensions
```

##### Change the dimensions to 1024 

```python
# Change the dimensions to 1024 for the "text-embedding-3-large" model
from langchain_openai import OpenAIEmbeddings

embeddings_1024 = OpenAIEmbeddings(
    model="text-embedding-3-large",
    dimensions=1024
)


text = "This is a practical example on OPENAI embedding"

query_result = embeddings_1024.embed_query(text)

len(query_result) # 1024 -> the length of the embedding vector for the "text-embedding-3-large" model with dimensions set to 1024 is 1024 dimensions
```

##### load text file and create embedding on it

```python
# Now let's load a text file using the TextLoader from langchain_community.document_loaders
from langchain_community.document_loaders import TextLoader

loader = TextLoader("sample.txt", encoding="utf8")
docs = loader.load()
docs
```

```python
# file is loaded, now let's split the documents into chunks

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)

final_documents = text_splitter.split_documents(docs)
final_documents
```

```python
# Now we have the documents split into chunks of 500 characters with an overlap of 50 characters. 
# We can now create embeddings for these chunks.

# embedding techniques are already defined above as embeddings_1024. We can now create a vector store using ChromaDB to store these embeddings.
```

```python
## Vector Embedding and Vector StoreDB
## for this let's use ChromaDB

from langchain_community.vectorstores import Chroma

# pass the document(final_documents) and embedding technique(embeddings_1024) to the Chroma vector store
db = Chroma.from_documents(final_documents, embeddings_1024)

db
```

```python
query = "The experts will enjoy the same treatment as their counterparts from State-owned firms, the report said."

# take this query and search for similar documents in the vector store using the similarity_search method
retrieved_result = db.similarity_search(query)

print(retrieved_result) # this will return the most similar documents to the query from the vector store
```
