#### Sentence Transformers on Hugging Face

- Hugging Face sentence-transformers is a Python framework for state-of-the-art sentence, text and image embeddings.
- One of the embedding models is used in the HuggingFaceEmbeddings class.
- We have also added an alias for SentenceTransformerEmbeddings for users who are more familiar with directly using the package.

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)
```

```python
text = "This is a testing document for embedding with hugging face"

query_result = embeddings.embed_query(text)

query_result
```

```python
print(len(query_result))
```

```python
doc_result = embeddings.embed_documents([text, "This is a testing document for embedding with hugging face"])

# embed_documents -> we can use multiple list of documents

doc_result[0]
```
